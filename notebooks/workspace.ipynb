`TypedDict` - Define the structure of a dictionary

Python dictionary can contain any `keys` and `values`. **TypeDict** describes which keys you expect and what types their values should have.

```python
from typing import TypedDict
```

```python
# Define the expected structure of a traveler dictionary
class Traveler(TypedDict):
    name: str
    destination: str
    days: int
```

**Why Annotated is useful in LangGraph?**

In langGraph, Annotated can attach instruction about how state updates should be handled.

Here:

- `list[str]` says that messages is a list of strings
- `Annotated[...., operator.add]` attaches the **operator.add** reducer to that field.
- LangGraph uses that reducer to determine how to combine updates to messages.

```python
from typing import TypedDict, Annotated
import operator

class Traveler(TypedDict):
    destination: str
    messages: Annotated[list[str], operator.add]
```

**operator.add** with lists

This is especially important for your LangGraph backend.

```python
import operator

old_messages = ["Hello"]
new_messages = ["plan a trip to japan"]

combined_messages = operator.add(old_messages, new_messages)

print(combined_messages)
```

For lists, **operator.add** concatenates the lists. It does not their individual elements numerically.

```python
type(combined_messages)
```

```python
from typing import TypedDict, Annotated
import operator

# Define the structure of the chatbot's shared state
class TravelerState(TypedDict):
    destination: str

    messages: Annotated[list[str], operator.add]


state: TravelerState = {
    "destination": "Japan",
    "messages": ["User: Plan a trip to Japan"]
}

# A later node returns an update to the messages field
new_update = {
    "messages": ["AI: Here is your 7-days itineary"]
}

# Demonstrate the reducer behavior directly
state["messages"] = operator.add(
    state["messages"],
    new_update["messages"]
)

print("Destination", state["destination"])
print("Conversation history:")

for message in state["messages"]:
    print(message)
```

**Quick interview answer**

- `TypedDict:` Defines the expected keys and value types of a dictionary.
- `Annotated:` Attaches metadata to a type. LangGraph can use this metadata to configure state reducers.
- `operator:` A Python module that exposes operations as functions. In LangGraph, operator.add is often used to append list-based state updates.

**Remember:** TypedDict defines the structure, Annotated attaches the reducer metadata, and operator.add provides the combining operation.
